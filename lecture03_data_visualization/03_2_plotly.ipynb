#### 1. 전세 보증금 계산 공식

월세에서 전세로 전환할 때는 법적 상한율이 정해져 있지 않아, 보통 양측의 합의나 시중 **전월세 전환율(5~5.5%)** 을 적용합니다.

##### 계산 공식

$$\text{전세 보증금} = \frac{\text{월세} \times 12}{\text{전환율}} + \text{기존 보증금}$$

---

##### 예시

> 보증금 **5,000만 원**에 월세 **70만 원**을 전세로 전환할 경우 (전환율 **5%** 적용)

| 단계 | 계산 | 결과 |
|------|------|------|
| ① 연간 월세 | 70만 원 × 12개월 | **840만 원** |
| ② 환산 보증금 | 840만 원 ÷ 0.05 | **1억 6,800만 원** |
| ③ 새로운 전세금 | 1억 6,800만 원 + 5,000만 원 | **2억 1,800만 원** |

```python
# 전월세 전환율 설정 (단위: 소수)
CONVERSION_RATE = 0.05  # 5%
def convert_to_jeonse(grfe, rtfe, rate=CONVERSION_RATE):
    """
    월세를 전세 보증금으로 환산하는 함수
    
    Parameters
    ----------
    grfe : float - 기존 보증금 (만원)
    rtfe : float - 월세 (만원)
    rate : float - 전월세 전환율 (기본값: 5%)
    
    Returns
    -------
    float - 환산 전세 보증금 (만원)
    
    공식: (월세 × 12) ÷ 전환율 + 기존 보증금
    """
    return (rtfe * 12) / rate + grfe
# 월세(RENT_SE == 'M') 행에 대해서만 전세 환산 보증금 계산
# RENT_SE 값은 실제 데이터에 맞게 조정하세요 ('B':전세, 'M':월세 또는 '1','2' 등)

seoul_rent_2026_new['JEONSE_CONVERTED'] = seoul_rent_2026_new.apply(
    lambda row: convert_to_jeonse(row['GRFE'], row['RTFE'])
    if row['RTFE'] > 0   # 월세가 있는 경우만 환산
    else row['GRFE'],    # 이미 전세인 경우 기존 보증금 그대로 사용
    axis=1
)
# 결과 확인
seoul_rent_2026_new[['RENT_SE', 'GRFE', 'RTFE', 'JEONSE_CONVERTED']].head(10)
```

```python
# 임대면적(결측없음) 제곱미터당 전세가 생성

# 제곱미터당 전세 환산 보증금 (만원/㎡)
seoul_rent_2026_new['JEONSE_PER_M2'] = (
    seoul_rent_2026_new['JEONSE_CONVERTED'] / seoul_rent_2026_new['RENT_AREA']
)
# 결과 확인
seoul_rent_2026_new[['RENT_AREA', 'JEONSE_CONVERTED', 'JEONSE_PER_M2']].head(10)
```

## Cloudflare Pages 발행 (소요 시간: 약 1분)

### 1️⃣ Cloudflare 프로젝트 생성
1. **[Cloudflare Dashboard](https://dash.cloudflare.com/)**에 로그인 후 **`Workers & Pages`** 메뉴로 이동합니다.
2. **`Create application`** $\rightarrow$ **`Pages`** 탭 선택 후 **`[Connect to Git]`** 버튼을 클릭합니다.
3. 저장소 목록에서 **`jslee-gses/seoul_rent_api`** 저장소를 선택합니다.

### 2️⃣ 빌드 설정 (Build Settings)
| 설정 항목 | 입력 값 |
| :--- | :--- |
| **Framework preset** | `None` |
| **Build command** | *(빈칸)* |
| **Build output directory** | `/` |

---

### 3️⃣ API Key 환경 변수 (Environment Variables) 등록
프로젝트 생성 완료 후, **`Settings`** $\rightarrow$ **`Environment variables`** 메뉴로 이동하여 아래 변수를 등록합니다.

* **Variable name**: `SEOUL_API_KEY`
* **Value**: `본인의 API키 입력`

> 💡 **완료**: 환경 변수 저장 후 **`Deployments`** 메뉴에서 **[Retry deployment]** (재배포)를 클릭하시면 서비스가 최종 등록됩니다!
