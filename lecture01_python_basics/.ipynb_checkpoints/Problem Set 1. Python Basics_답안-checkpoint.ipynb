#### 1. 자료형의 이해 : python으로 생일파티 초대 문구 만들기

(1) 다음 변수에 여러분들의 이름, 태어난 달, 태어난 날짜, 나이, 고향을 정의하여라. 단, 표기된 자료형에 맞게 선언하여라

```python
# 코드 작성
name = "최준혁"
month = 6
date = 21
age = 27
```

(2) name 문자열에서 indexing(대괄호)을 사용하여 성씨만 추출하고, 이를 first 변수에 선언하여라. 그리고 다음 문구를 print 함수를 이용해 출력하여라

```python
# 코드 작성
first = name[0]
print('%s씨 가문의 자랑!, %s의 생일파티에 당신을 초대합니다' % (first, name))
print(first +'씨 가문의 자랑!, '+ name + '의 생일파티에 당신을 초대합니다')
```

(3) 입력한 age 변수와 숫자 연산자를 활용하여, 나잇대 숫자로 변환하여라 (20, 30, 40 등등) \
ex) 나이가 17세라면, 10을 출력한다

```python
# 코드 작성
a = age // 10
b = a*10
print(b)
```

(4) 위 결과를 바탕으로 다음 문구를 출력하여라. 이 때 줄 바꿈까지 완료하여라

```python
# 코드 작성
b = (age // 10) * 10
print('화려한 %d대 신나게 놀아보자~~ 언제?\n- %d월 %d일 저녁에 %s의 집에서 -'% (b,month,date,name))
```

- note\
    숫자와 문자열을 합칠 경우, 숫자를 문자열로 변환한 후 합칠 수 있습니다.

```python
a = 1600
b = "원"
print(str(a)+b)    #str() 함수를 통해 숫자를 문자로, int() 함수를 통해 문자를 숫자로. 단 int함수는 숫자 문자열에만 적용 가능
```

#### 2. list와 반복문 : 홀짝게임

다음 변수를 선언하고 물음에 답하여라

```python
numbers = [12,18,11,54,13]
```

(1) numbers list에서 11을 지우고, 마지막에 31을 추가하여라

```python
#답안 작성
del numbers[2]
print(numbers)
numbers.append(31)
print(numbers)
```

(2) 먼저 단순한 for문을 통해 numbers의 숫자를 차례대로 출력하여라

```python
#답안 작성
for i in numbers:
    print(i)
```

(3) for문 안에 if문을 중첩하여, numbers의 숫자가 홀수일 때만 출력하여라

```python
#답안 작성
for i in numbers:
    if i % 2 == 1:
        print(i)
```

(4) 위에서 만든 if문을 발전시켜, numbers의 숫자와 홀짝 여부를 차례대로 출력하여라

ex)\
12 짝수\
18 짝수\
...

```python
#답안 작성
for i in numbers:
    if i % 2 == 0:
        print(i, "짝수")
    else:
        print(i, "홀수")
```

(5) time 라이브러리의 sleep 메서드를 이용하면, 코드 실행을 정해진 시간동안 일시정지 할 수 있다. 다음 코드를 실행시켜 확인해보자

```python
import time

print ("5초동안 중지")
time.sleep(5)
print("일어나기")
```

time.sleep()을 이용하여 숫자와 홀짝 여부를 알기 전 맞출 수 있는 시간을 주어라. 또한, 그 동안 "맞춰보세요"라는 안내 문구도 추가하여라

ex)\
맞춰보세요\
(5초 휴식...)\
12 짝수


```python
#답안 작성
for i in numbers:
    if i % 2 == 0:
        print("맞춰보세요")
        time.sleep(2)
        print(i, "짝수")
    else:
        print("맞춰보세요")
        time.sleep(2)
        print(i, "홀수")
```

#### 3. 함수 : 1~100까지의 합 구하기

(1) 먼저 입력 값을 받아 1을 더해 출력하는 간단한 함수 addition을 선언하여라. 또한 함수가 잘 작동하는지 확인하여라

```python
#답안 작성
def addition(a):
    return a+1

addition(3)
```

(2) for문을 이용해 total 변수에 0부터 100까지 차례로 더하여라.

```python
total = 0
#답안작성
for i in range(101):
     total = total + i
print(total)
```

(3) 0부터 입력값 a까지 합을 구하는 함수를 정의하고, 이를 이용하여 0부터 100까지의 합을 구하여라

```python
#답안작성
def addition(a):
    total = 0
    for i in range(a+1):
        total = total+i
    return total

addition(100)
```

#### 딕셔너리 : 라면 자판기

- 라면 자판기에는 총 4가지의 라면을 팔고 있습니다. (신라면, 삼양라면, 비빔면, 짜파게티)
- 신라면은 1000원, 삼양라면은 800원, 비빔면은 1200원, 짜파게티는 1500원입니다.

(1) 위에서 언급된 라면의 품목과 가격을 딕셔너리 변수로 선언하여라

```python
#답안작성
ramen = {'신라면' : 1000, '삼양라면' : 800, '비빔면' : 1200, '짜파게티' : 1500}
```

(2) input 함수를 이용하여 돈을 입력받고, 사고싶은 물건, 개수를 입력받아라. 이를 각각 money, product, number 변수로 할당하여라\
주) input 함수는 문자열(string)형태로 입력받으므로, 필요하다면 자료형을 변환시켜야 한다

```python
#답안작성
money = input('돈 입력 : ')
product = input('사고싶은 품목 : ')
number = input('사고싶은 개수 : ')
```

(3) 입력받은 품목(product), 개수(number) 정보를 바탕으로 하여, 전체 가격과 거스름돈을 계산하여라

```python
#답안작성
money = int(money)
number = int(number)
total = ramen[product]*number
remain = money-total
```

(4) if 문을 활용하여, 만약 구매가 가능하다면 다음 문구와 거스름돈을 출력하여라. 만약 불가능하다면 잔돈 부족 문구를 출력하여라

```python
#답안 작성
if remain > 0:
    print('구매가 완료되었습니다. 거스름돈 : %d원' %(remain))
else:
    print('잔액이 부족합니다')
```
