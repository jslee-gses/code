- 소수의 예: 2, 3, 5, 7, 11,...

```python
# Hint 3
# 입력된 정수를 2부터 "자기자신-1"까지 차례로 나누어 나머지를 rest라는 리스트에 저장
x=5
rest = [x % i for i in range(2, x)]
rest
```

```python
# Hint 4
# rest에 0이 포함되면 1과 자기 자신 이외에 다른 수를 약수로 가지므로 소수가 아님
if 0 in rest: 
    print("%d is not a prime number." % x)
else: 
    print("%d is a prime number." % x)
```

## 소수판별기 함수를 작성하시오 

```python
# Answer
def prime(x):
    if x < 2:
        print("%d is not a prime number." % x)
    else: 
        rest = [x % i for i in range(2, x)]
        if 0 in rest: 
            print("%d is not a prime number." % x)
        else: 
            print("%d is a prime number." % x)
        
prime(13)
```

## 정수를 입력받는 함수와 소수판별기 함수를 합하시오

```python
# Answer

def user_input():
    x = int(input("Pleae enter a integer :" ))
    print("your number is:", x)
    return x

def prime(x):
    if x < 2:
        print("%d is not a prime number." % x)
    else: 
        rest = [x % i for i in range(2, x)]
        if 0 in rest: 
            print("%d is not a prime number." % x)
        else: 
            print("%d is a prime number." % x)
    
def main():
    num = user_input()
    prime(num)
    
# 호출
main()
```
