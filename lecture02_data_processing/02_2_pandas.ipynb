## Mutating Categorical Data : Map and Cut


```python
flight_data
```

```python
flight_data.carrier.unique()
```

```python
# 1. 항공사 그룹 리스트 정의 (가독성 향상)
class_A = ['UA', 'AA', 'B6', 'DL', 'EV', 'MQ', 'US']
class_B = ['WN', 'VX', 'FL', 'AS', '9E', 'F9', 'HA', 'YV', 'OO']

# 2. 조건 및 결과 지정
conditions = [
    flight_data['carrier'].isin(class_A),
    flight_data['carrier'].isin(class_B)
]
choices = ['A', 'B']

# 3. 매핑 실행 (리스트에 없는 신규 항공사는 'Other'로 지정)
flight_data['carrier_class'] = np.select(conditions, choices, default='Other')
flight_data[['carrier', 'carrier_class']].sample(n=10)
```

```python
# flight_data['cacarrier_class']=='UA'이면,  'carricarrier_class' 를 'B'로 변환해서 수정

flight_data.loc[flight_data['carrier']=='UA', 'carrier_class'] = 'B'


flight_data[['carrier', 'carrier_class']].sample(n=10)
```

```python
flight_data.distance.describe()
```

```python
# Caterorical variable from Continuous Variable

# 0 포함(include_lowest=True) 및 5000 초과 거리 무한대(np.inf) 커버
bins = [0, 502, 872, 1389, np.inf]
labels = ["very_short", "short", "long", "very_long"]
flight_data['distance_class'] = pd.cut(
    flight_data['distance'], 
    bins=bins, 
    labels=labels,
    include_lowest=True  # 0 값도 안전하게 구간에 포함
)
flight_data[['distance', 'distance_class']].sample(n=10)
```

## 연습문제 02

* \[Step 1\]: Filter out all flights less than 10 hours
* \[Step 2\]: Create a new column, `speed`, using the formula [distance / (air time * 60)]
* \[Step 3\]: Calculate the mean speed for flights originating from each airport
* \[Step 4\]: Sort the result by mean speed in descending order

```python
# 1. hour가 10 초과인 데이터 필터링 (.copy()로 경고 방지)
filtered_data = flight_data[flight_data['hour'] > 10].copy()
# 2. 속도(mph) 계산 및 컬럼 추가
filtered_data['speed'] = (filtered_data['distance'] / filtered_data['air_time']) * 60
# 3. origin별 평균 속도 계산 및 정렬
result = (
    filtered_data.groupby('origin')['speed']
    .agg(mean_speed='mean')
    .reset_index()
    .sort_values('mean_speed', ascending=False)
)
result
```

## From Wide to Long

```python
# 1. 가로로 넓은 형태의 예시 데이터프레임 생성
data = {
    'id': [1, 2],
    'A2020': [100, 200],
    'A2021': [120, 220],
    'B2020': [15, 25],
    'B2021': [18, 28]
}
df = pd.DataFrame(data)
df
```

```python
# 2. pd.wide_to_long으로 변환
# - stubnames=["A", "B"]: 변환할 컬럼명의 공통 접두사
# - i="id": 각 행을 구분하는 식별자
# - j="year": 접두사 뒤에 붙은 숫자가 새로 들어갈 컬럼명
df_long = pd.wide_to_long(df, stubnames=["A", "B"], i="id", j="year").reset_index()

df_long
```

```python
# 1. 다중 식별자를 가진 Wide 형태 데이터프레임 생성
# famid(가족ID), birth(출생순위), ht1(1세 때 키), ht2(2세 때 키)
df = pd.DataFrame({
    'famid': [1, 1, 2, 2],
    'birth': [1, 2, 1, 2],
    'ht1': [50, 48, 52, 51],
    'ht2': [80, 76, 82, 81]
})
df
```

```python
# 2. pd.wide_to_long 실행
# - stubnames='ht': 변환할 컬럼명의 공통 접두사 ('ht')
# - i=['famid', 'birth']: 다중 식별자 컬럼 리스트
# - j='age': ht1, ht2의 숫자(1, 2)가 새로 들어갈 컬럼명
result = pd.wide_to_long(df, stubnames='ht', i=['famid', 'birth'], j='age').reset_index()
result
```

## Join

```python
# 복사본 생성
flight_data_reset  = flight_data

flight_data_reset
```

```python

# index를 활용해서 id 컬럼 생성

flight_data_reset['id'] = range(1, len(flight_data_reset) + 1)

flight_data_reset
```

```python
# id 컬럼을 공유하는 두 개의 데이터 프레임 생성
flight_distance = flight_data_reset.loc[flight_data_reset['carrier'].isin(['DL','AA']), ['id','carrier','distance']] # DL, AA
flight_time = flight_data_reset.loc[flight_data_reset['carrier'].isin(['AA']), ['id','carrier','air_time']] #  AA 만
```

```python
flight_distance
```

```python
flight_time
```

병합 방식은 크게 inner, outer, left, right 방식이 있습니다.
* inner join 방식 : 좌측 table과 우측 table의 기준열에서 key 값이 모두 존재하는 경우에만 병합
* outer join 방식 : 좌측 table과 우측 table의 기준열에서 key 값이 모두 존재하지 않아도 병합
* left join 방식 : 좌측 table의 기준열만 고려하여 병합
* right join 방식 : 우측 table의 기준열만 고려하여 병합

여기서는 많이 사용하는 inner join, left join에 대해서만 배우겠습니다.

<img src="https://static.wixstatic.com/media/2f98e9_c5999377496a4e399bbc042ffe3280fc~mv2.jpg" width="600" height="400" />

```python
# 왼쪽의 기준 Table을 먼저 쓰고, 오른쪽의 병합 Table을 다음에 씁니다. left_on과 right_on은 각각 Table의 기준열입니다.
#inner join 방식입니다. inner join이  default 설정이기 때문에 how = 'inner' parameter를 넣지 않아도 됩니다.

flight_inner = pd.merge(flight_distance, flight_time, left_on='id', right_on="id", how='inner')

flight_inner
```

flight_distance은 80839개의 행을 가지고 있었으나, inner join은 'flight_distance' 'flight_time'에 공통적으로 포함된 값들만 살리고, 이들을 기준으로 데이터를 병합합니다. 따라서 flight_distance에서 carrier가 AA가 아닌 행들은 모두 삭제됩니다.

```python
#left join 방식입니다.
flight_left = pd.merge(flight_distance, flight_time, left_on='id', right_on="id", how='left')

flight_left
```

left join은 왼쪽의 기준테이블의 모든행을 유지합니다. 따라서 flight_distance은 80839개의 행을 모두 유지합니다. 기준테이블과 결합테이블 간에 일치하지 않는 행은 결측치가 입력됩니다.

## Stacking
