### 🌐 API(Application Programming Interface)란 무엇인가?
#### 1. API의 정의
* **API**는 **Application Programming Interface**의 약자로, 서로 다른 소프트웨어 프로그램이나 시스템이 **서로 데이터를 주고받을 수 있게 연결해 주는 창구(매개체)**입니다.
* 데이터베이스나 서버에 직접 접근하지 않고, 규격화된 약속(인터페이스)을 통해 안전하게 필요한 데이터만 요청하고 응답받을 수 있습니다.
---
#### 2. API를 이해하는 가장 쉬운 비유: 🍽️ 식당의 점원 (Waiter)
| 구분 | 역할 비유 | 실제 컴퓨터 환경 |
| :--- | :--- | :--- |
| **손님 (Client)** | 메뉴를 보고 음식을 주문함 | **내 컴퓨터 / 파이썬 프로그램 / 브라우저** |
| **점원 (API)** | 주문을 주방에 전달하고, 음식을 손님에게 가져다줌 | **API (Application Programming Interface)** |
| **주방 (Server)** | 데이터(음식)를 보관하고 요리(처리)함 | **원격 데이터베이스 / 서버** |
> 💡 우리가 직접 주방(서버)에 들어가지 않아도, **점원(API)**에게 올바른 메뉴(요청)를 말하면 정해진 방식(응답)으로 음식(데이터)을 받아볼 수 있습니다.
---
#### 3. 웹 API의 핵심 구조: 요청(Request)과 응답(Response)
API와의 통신은 크게 **두 가지 단계**로 이루어집니다.
1. **요청 (Request)**: 클라이언트가 서버에게 특정 데이터를 달라고 보냄
   * **URL (주소)**: 데이터를 요청할 위치
   * **인증키 (API Key)**: 접근 권한 확인용 열쇠
   * **파라미터 (Parameter)**: 원하는 데이터 조건 (예: 연도, 자치구, 개수 등)
2. **응답 (Response)**: 서버가 처리 결과를 클라이언트에게 전달함
   * 주로 **JSON** 또는 **XML** 형식을 사용합니다.
---
#### 4. 데이터 교환의 표준 형식: JSON (JavaScript Object Notation)
현재 대부분의 웹 API는 데이터를 **JSON** 형식으로 제공합니다.
* Python의 **사전(Dictionary)** 구조(`{ Key : Value }`)와 매우 유사합니다.
* 읽고 쓰기 쉽고 가벼워서, 파이썬의 `pandas` 라이브러리를 통해 **표(DataFrame)** 형태로 손쉽게 변환할 수 있습니다.
```json
{
  "name": "홍길동",
  "age": 25,
  "city": "서울"
}

````python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
# .env 파일 로드
load_dotenv()
# 환경 변수에서 API 키 읽기
API_KEY = os.getenv("SEOUL_API_KEY")
TYPE = "json"
SERVICE = "tbLnOpendataRentV"
all_rows = []
start_idx = 1
step = 1000
max_data = 5000  # 수집할 총 데이터 건수 예시
while start_idx <= max_data:
    end_idx = start_idx + step - 1
    url = f"http://openapi.seoul.go.kr:8088/{API_KEY}/{TYPE}/{SERVICE}/{start_idx}/{end_idx}/"
    
    res = requests.get(url)
    data = res.json()
    
    if SERVICE in data:
        rows = data[SERVICE]['row']
        all_rows.extend(rows)
        start_idx += step
    else:
        # 오류 또는 데이터 없음 시 종료
        print("결과 없음 또는 오류 발생:", data.get('RESULT', {}).get('MESSAGE'))
        break
df_total = pd.DataFrame(all_rows)
print(f"총 {len(df_total)}건 수집 완료")
df_total.head()
````

접수연도가 2026년인 데이터만 1,000건씩 분할하여 전체 수집할 수 있도록 수정

````python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
# 1. .env 파일 로드 및 API 키 읽기
load_dotenv()
API_KEY = os.getenv("SEOUL_API_KEY")
# 2. API 기본 정보 및 수집 조건 설정
TYPE = "json"
SERVICE = "tbLnOpendataRentV"
RCPT_YR = "2026"  # 접수연도 2026년 조건 추가
all_rows = []
start_idx = 1
step = 1000
total_count = None
print(f"[{RCPT_YR}년 서울시 부동산 전월세가 데이터 수집 시작]")
# 3. 데이터 반복 수집 (Paging)
while True:
    end_idx = start_idx + step - 1
    
    # URL 맨 뒤에 {RCPT_YR} 파라미터 추가
    url = f"http://openapi.seoul.go.kr:8088/{API_KEY}/{TYPE}/{SERVICE}/{start_idx}/{end_idx}/{RCPT_YR}/"
    
    res = requests.get(url)
    data = res.json()
    
    if SERVICE in data:
        # 첫 호출 시 2026년 전체 데이터 건수 자동 감지
        if total_count is None:
            total_count = data[SERVICE]['list_total_count']
            print(f"총 수집 대상 데이터: {total_count:,}건")
        
        rows = data[SERVICE]['row']
        all_rows.extend(rows)
        print(f"수집 진행 중: {len(all_rows):,} / {total_count:,} 건 완료")
        
        # 다음 1,000건으로 인덱스 이동
        start_idx += step
        
        # 전체 데이터 건수 이상 수집 시 루프 종료
        if len(all_rows) >= total_count:
            break
    else:
        # 오류 또는 데이터가 더 이상 없는 경우 종료
        msg = data.get('RESULT', {}).get('MESSAGE', '데이터 없음')
        print(f"수집 종료/중단: {msg}")
        break
# 4. DataFrame 변환 및 확인
df_total = pd.DataFrame(all_rows)
print(f"\n✅ 최종 {len(df_total):,}건 수집 완료!")
df_total.head()
````

## Cloudflare Pages 발행 (소요 시간: 약 1분)

### 1️⃣ Cloudflare 프로젝트 생성
1. **[Cloudflare Dashboard](https://dash.cloudflare.com/)**에 로그인 후 **`Workers & Pages`** 메뉴로 이동합니다.
2. **`Create application`** $\rightarrow$ **`Pages`** 탭 선택 후 **`[Connect to Git]`** 버튼을 클릭합니다.
3. 저장소 목록에서 **`jslee-gses/seoul_rent_api`** 저장소를 선택합니다.

### 2️⃣ 빌드 설정 (Build Settings)
| 설정 항목 | 입력 값 |
| :--- | :--- |
| **Framework preset** | `None` |
| **Build command** | *(빈칸)* |
| **Build output directory** | `/` |

---

### 3️⃣ API Key 환경 변수 (Environment Variables) 등록
프로젝트 생성 완료 후, **`Settings`** $\rightarrow$ **`Environment variables`** 메뉴로 이동하여 아래 변수를 등록합니다.

* **Variable name**: `SEOUL_API_KEY`
* **Value**: `본인의 API키 입력`

> 💡 **완료**: 환경 변수 저장 후 **`Deployments`** 메뉴에서 **[Retry deployment]** (재배포)를 클릭하시면 서비스가 최종 등록됩니다!
